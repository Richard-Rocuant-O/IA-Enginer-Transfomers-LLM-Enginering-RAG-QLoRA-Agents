```python
# Conectar a Google y Ollama; Ollama corre en tu máquina y no necesita clave API

# openai = OpenAI()  # No necesitamos OpenAI, usamos Gemini en su lugar

anthropic_url = "https://api.anthropic.com/v1/"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
ollama_url = "http://localhost:11434/v1"

# anthropic = OpenAI(api_key=anthropic_api_key, base_url=anthropic_url)
gemini = OpenAI(api_key=google_api_key, base_url=gemini_url)
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
```

```python
# Vamos a envolver una llamada a Gemini en una función simple

system_message = "Eres un asistente útil, la fecha de hoy es 08-10-2024, y tu conocimiento se corta en 2024-06. Responde de manera concisa y clara."

def message_gemini(prompt):
    messages = [{"role": "system", "content": system_message}, {"role": "user", "content": prompt}]
    response = gemini.chat.completions.create(model="gemini-3.1-flash-lite", messages=messages)
    return response.choices[0].message.content
```

```python
# Esto puede revelar la "fecha de corte de entrenamiento" o la fecha más reciente en los datos de entrenamiento

message_gemini("¿Cuál es la fecha de hoy?")
```

## Forzar modo oscuro

Gradio aparece en modo claro u oscuro dependiendo de la configuración del navegador y la computadora. Hay una forma de forzar que gradio aparezca en modo oscuro, pero Gradio lo desaconseja ya que debería ser una preferencia del usuario (particularmente por razones de accesibilidad). Pero si deseas forzar el modo oscuro para tus pantallas, aquí es cómo hacerlo.

```python
# Define esta variable y luego pasa js=force_dark_mode cuando crees la Interface

force_dark_mode = """
function refresh() {
    const url = new URL(window.location);
    if (url.searchParams.get('__theme') !== 'dark') {
        url.searchParams.set('__theme', 'dark');
        window.location.href = url.href;
    }
}
"""
gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never", js=force_dark_mode).launch()
```

```python
# Vamos a usar Markdown
# ¿Te preguntas por qué importa establecer system_message si no se hace referencia en el código abajo?
# Estoy aprovechando que system_message sea una variable global, utilizada en la función message_gemini (revisa)
# No es una gran práctica de ingeniería de software, ¡pero bastante común durante R&D en Jupyter Lab!

system_message = "Eres un asistente útil que responde en markdown sin bloques de código"

message_input = gr.Textbox(label="Tu mensaje:", info="Ingresa un mensaje para Gemini", lines=7)
message_output = gr.Markdown(label="Respuesta:")

view = gr.Interface(
    fn=message_gemini,
    title="Gemini", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Explica la arquitectura Transformer a una persona lega",
        "Explica la arquitectura Transformer a un ingeniero de IA aspirante",
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
# Vamos a crear una llamada que transmita resultados
# Si quieres actualizar tu conocimiento sobre Generadores (la palabra clave "yield"),
# Por favor revisa la guía de Python Intermedio en la carpeta de guías

def stream_gemini(prompt):
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
      ]
    stream = gemini.chat.completions.create(
        model='gemini-3.1-flash-lite',
        messages=messages,
        stream=True
    )
    result = ""
    for chunk in stream:
        result += chunk.choices[0].delta.content or ""
        yield result
```

```python
message_input = gr.Textbox(label="Tu mensaje:", info="Ingresa un mensaje para Gemini", lines=7)
message_output = gr.Markdown(label="Respuesta:")

view = gr.Interface(
    fn=stream_gemini,
    title="Gemini", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Explica la arquitectura Transformer a una persona lega",
        "Explica la arquitectura Transformer a un ingeniero de IA aspirante",
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
def stream_ollama(prompt):
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
      ]
    stream = ollama.chat.completions.create(
        model='gemma3:270m',
        messages=messages,
        stream=True
    )
    result = ""
    for chunk in stream:
        result += chunk.choices[0].delta.content or ""
        yield result
```

```python
message_input = gr.Textbox(label="Tu mensaje:", info="Ingresa un mensaje para Gemma 3 (Ollama)", lines=7)
message_output = gr.Markdown(label="Respuesta:")

view = gr.Interface(
    fn=stream_ollama,
    title="Ollama",
    inputs=[message_input],
    outputs=[message_output],
    examples=[
        "Explica la arquitectura Transformer a una persona lega",
        "Explica la arquitectura Transformer a un ingeniero de IA aspirante",
        ],
    flagging_mode="never"
    )
view.launch()
```

## Y ahora convirtiéndose en sofisticado

Recuerda revisar la Guía de Python Intermedio si no estás seguro sobre generadores y "yield"

```python
def stream_model(prompt, model):
    if model=="Gemini":
        result = stream_gemini(prompt)
    elif model=="Ollama":
        result = stream_ollama(prompt)
    else:
        raise ValueError("Modelo desconocido")
    yield from result
```

```python
message_input = gr.Textbox(label="Tu mensaje:", info="Ingresa un mensaje para el LLM", lines=7)
model_selector = gr.Dropdown(["Gemini", "Ollama"], label="Selecciona modelo", value="Gemini")
message_output = gr.Markdown(label="Respuesta:")

view = gr.Interface(
    fn=stream_model,
    title="LLMs",
    inputs=[message_input, model_selector],
    outputs=[message_output],
    examples=[
            ["Explica la arquitectura Transformer a una persona lega", "Gemini"],
            ["Explica la arquitectura Transformer a un ingeniero de IA aspirante", "Ollama"]
        ],
    flagging_mode="never"
    )
view.launch()
```

```python
import sys
sys.path.append("../week1")
from scraper import fetch_website_contents
```

```python

# De nuevo esto es una mentalidad típicamente Experimental - estoy cambiando la variable global que usamos arriba:

system_message = """
Eres un asistente que analiza el contenido de la página de inicio del sitio web de una empresa
y crea un folleto breve sobre la empresa para clientes potenciales, inversores y nuevos talentos.
Responde en markdown sin bloques de código.
"""
```

```python
def stream_brochure(company_name, url, model):
    yield ""
    prompt = f"Por favor genera un folleto de empresa para {company_name}. Aquí está su página de inicio:\n"
    prompt += fetch_website_contents(url)
    if model=="Gemini":
        result = stream_gemini(prompt)
    elif model=="Ollama":
        result = stream_ollama(prompt)
    else:
        raise ValueError("Modelo desconocido")
    yield from result
```

```python
name_input = gr.Textbox(label="Nombre de empresa:")
url_input = gr.Textbox(label="URL de página de inicio incluyendo http:// o https://")
model_selector = gr.Dropdown(["Gemini", "Ollama"], label="Selecciona modelo", value="Gemini")
message_output = gr.Markdown(label="Respuesta:")

view = gr.Interface(
    fn=stream_brochure,
    title="Generador de Folletos",
    inputs=[name_input, url_input, model_selector],
    outputs=[message_output],
    examples=[
            ["Hugging Face", "https://huggingface.co", "Gemini"],
            ["Edward Donner", "https://edwarddonner.com", "Ollama"]
        ],
    flagging_mode="never"
    )
view.launch()
```
