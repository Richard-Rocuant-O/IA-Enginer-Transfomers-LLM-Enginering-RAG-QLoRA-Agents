```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
# Cargar variables de entorno de un archivo llamado .env
# Imprimir los prefijos de clave para ayudar con cualquier depuración

load_dotenv(override=True)
google_api_key = os.getenv('GOOGLE_API_KEY')

if google_api_key:
    print(f"La clave de API de Google existe y comienza con {google_api_key[:8]}")
else:
    print("La clave de API de Google no está configurada")
```

```python
# Inicializar

gemini = OpenAI(api_key=google_api_key, base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
MODEL = 'gemini-3.1-flash-lite'
```

```python
# De nuevo, voy a estar en modo científico y cambiar esta variable global durante el laboratorio

system_message = "Eres un asistente útil"
```

## ¡OK! ¡Escribamos un callback de chat un poco mejor!

```python

def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    response = gemini.chat.completions.create(model=MODEL, messages=messages)
    return response.choices[0].message.content
```

```python
gr.ChatInterface(fn=chat, type="messages").launch()
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    stream = gemini.chat.completions.create(model=MODEL, messages=messages, stream=True)
    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

```python
gr.ChatInterface(fn=chat, type="messages").launch()
```

## ¡OK, sigamos!

Usando un mensaje del sistema para agregar contexto y dar una respuesta de ejemplo... esto es "one shot prompting" otra vez

```python
system_message = "Eres un asistente útil en una tienda de ropa. Debes intentar animar \
suavemente al cliente a probar artículos que están en oferta. Los sombreros tienen 60% de descuento, y la mayoría \
de los demás artículos tienen 50% de descuento. \
Por ejemplo, si el cliente dice 'Quiero comprar un sombrero', \
puedes responder algo como: '¡Maravilloso - tenemos muchísimos sombreros, incluidos varios que forman parte de nuestro evento de rebajas.'\
Anima al cliente a comprar sombreros si no está seguro de qué llevar."
```

```python
gr.ChatInterface(fn=chat, type="messages").launch()
```

```python
system_message += "\nSi el cliente pide zapatos, debes responder que los zapatos no están en oferta hoy, \
¡pero recuérdale que mire los sombreros!"
```

```python
gr.ChatInterface(fn=chat, type="messages").launch()
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    relevant_system_message = system_message
    if 'belt' in message.lower():
        relevant_system_message += " La tienda no vende cinturones; si te piden cinturones, asegúrate de señalar otros artículos en oferta."
    
    messages = [{"role": "system", "content": relevant_system_message}] + history + [{"role": "user", "content": message}]

    stream = gemini.chat.completions.create(model=MODEL, messages=messages, stream=True)

    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

```python
gr.ChatInterface(fn=chat, type="messages").launch()
```
