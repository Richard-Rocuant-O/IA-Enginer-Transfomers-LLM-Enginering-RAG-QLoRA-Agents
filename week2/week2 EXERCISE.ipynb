# Ejercicio adicional de fin de semana - semana 2

Ahora usa todo lo que aprendiste en la semana 2 para construir un prototipo completo del asistente de preguntas y respuestas técnicas que hiciste en el ejercicio de la semana 1.

Debe incluir una interfaz con Gradio, streaming, uso del prompt de sistema para añadir experiencia y la capacidad de cambiar entre modelos. ¡Puntos extra si logras demostrar el uso de una herramienta!

Si te animas, prueba a añadir entrada de audio para hablarle y que responda con audio. ChatGPT o Claude pueden ayudarte, o escríbeme por correo si tienes preguntas.

Publicaré una solución completa aquí pronto, a menos que alguien se me adelante...

Hay muchísimas aplicaciones comerciales para esto: desde un tutor de idiomas, hasta una solución de onboarding para empresas o un compañero de IA para un curso (¡como este!). Tengo muchas ganas de ver tus resultados.
