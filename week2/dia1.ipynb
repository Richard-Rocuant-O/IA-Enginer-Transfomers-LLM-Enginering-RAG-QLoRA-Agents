## Y ahora para un poco de diversión - ¡una conversación adversarial entre Chatbots!

Ya estás familiarizado con los prompts organizados en listas como:

```
[
    {"role": "system", "content": "mensaje del sistema aquí"},
    {"role": "user", "content": "prompt del usuario aquí"}
]
```

De hecho, esta estructura se puede usar para reflejar un historial de conversación más largo:

```
[
    {"role": "system", "content": "mensaje del sistema aquí"},
    {"role": "user", "content": "primer prompt del usuario aquí"},
    {"role": "assistant", "content": "la respuesta del asistente"},
    {"role": "user", "content": "el nuevo prompt del usuario"},
]
```

Y podemos usar este enfoque para entablar una interacción más larga con historial.


````python
# Hagamos una conversación entre GPT-4.1-mini y Claude-haiku-4.5
# Estamos usando versiones baratas de modelos para que los costos sean mínimos

gpt_model = "gpt-4.1-mini"
claude_model = "claude-haiku-4-5"

gpt_system = "Eres un chatbot que es muy argumentativo; \
desacuerdas con cualquier cosa en la conversación y cuestionas todo, de una manera sarcástica."

claude_system = "Eres un chatbot muy educado y cortés. Intentas estar de acuerdo con \
todo lo que dice la otra persona, o encontrar terreno común. Si la otra persona es argumentativa, \
intentas calmarla y seguir charlando."

gpt_messages = ["Hola allá"]
claude_messages = ["Hola"]
````

````python
def call_gpt():
    messages = [{"role": "system", "content": gpt_system}]
    for gpt, claude in zip(gpt_messages, claude_messages):
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": claude})
    response = openai.chat.completions.create(model=gpt_model, messages=messages)
    return response.choices[0].message.content
````

````python
call_gpt()
````

````python
def call_claude():
    messages = [{"role": "system", "content": claude_system}]
    for gpt, claude_message in zip(gpt_messages, claude_messages):
        messages.append({"role": "user", "content": gpt})
        messages.append({"role": "assistant", "content": claude_message})
    messages.append({"role": "user", "content": gpt_messages[-1]})
    response = anthropic.chat.completions.create(model=claude_model, messages=messages)
    return response.choices[0].message.content
````

````python
call_claude()
````

````python
call_gpt()
````

````python
gpt_messages = ["Hi there"]
claude_messages = ["Hi"]

display(Markdown(f"### GPT:\n{gpt_messages[0]}\n"))
display(Markdown(f"### Claude:\n{claude_messages[0]}\n"))

for i in range(5):
    gpt_next = call_gpt()
    display(Markdown(f"### GPT:\n{gpt_next}\n"))
    gpt_messages.append(gpt_next)
    
    claude_next = call_claude()
    display(Markdown(f"### Claude:\n{claude_next}\n"))
    claude_messages.append(claude_next)
````

## Ejercicio adicional

También podrías intentar reemplazar uno de los modelos con un modelo de código abierto corriendo con Ollama.


````python
# --- Personalidades ---
alex_system = """
Eres Alex, un chatbot que es muy argumentativo; desacuerdas con cualquier cosa en la conversación 
y cuestionas todo, de una manera sarcástica.
Estás en una conversación con Blake y Charlie.
"""

blake_system = """
Eres Blake, un chatbot muy educado y cortés. Intentas estar de acuerdo con todo lo que dicen los demás, 
o encontrar terreno común. Si alguien es argumentativo, intentas calmarlo.
Estás en una conversación con Alex y Charlie.
"""

charlie_system = """
Eres Charlie, un chatbot curioso y analítico. Te gusta hacer preguntas y profundizar en los temas 
con datos o ejemplos concretos.
Estás en una conversación con Alex y Blake.
"""

# --- Historial como texto plano, no como lista de roles ---
conversation = []  # cada elemento: ("Nombre", "lo que dijo")

def format_conversation():
    return "\n".join(f"{nombre}: {texto}" for nombre, texto in conversation)

def call_alex():
    user_prompt = f"""
Eres Alex, en conversación con Blake y Charlie.
La conversación hasta ahora es la siguiente:
{format_conversation()}
Ahora con esto, responde con lo que te gustaría decir a continuación, como Alex.
"""
    response = openai.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {"role": "system", "content": alex_system},
            {"role": "user", "content": user_prompt}
        ]
    )
    return response.choices[0].message.content

def call_blake():
    user_prompt = f"""
Eres Blake, en conversación con Alex y Charlie.
La conversación hasta ahora es la siguiente:
{format_conversation()}
Ahora con esto, responde con lo que te gustaría decir a continuación, como Blake.
"""
    response = anthropic.chat.completions.create(
        model="claude-haiku-4-5",
        messages=[
            {"role": "system", "content": blake_system},
            {"role": "user", "content": user_prompt}
        ]
    )
    return response.choices[0].message.content

def call_charlie():
    user_prompt = f"""
Eres Charlie, en conversación con Alex y Blake.
La conversación hasta ahora es la siguiente:
{format_conversation()}
Ahora con esto, responde con lo que te gustaría decir a continuación, como Charlie.
"""
    response = gemini.chat.completions.create(
        model="gemini-3.1-flash-lite",
        messages=[
            {"role": "system", "content": charlie_system},
            {"role": "user", "content": user_prompt}
        ]
    )
    return response.choices[0].message.content

# --- Mensajes iniciales ---
conversation.append(("Alex", "Hola a todos."))
conversation.append(("Blake", "¡Hola! Qué gusto estar aquí."))
conversation.append(("Charlie", "Hola, tengo curiosidad por ver de qué vamos a hablar."))

for nombre, texto in conversation:
    display(Markdown(f"### {nombre}:\n{texto}\n"))

# --- Bucle de 5 rondas ---
for i in range(5):
    alex_next = call_alex()
    display(Markdown(f"### Alex:\n{alex_next}\n"))
    conversation.append(("Alex", alex_next))

    blake_next = call_blake()
    display(Markdown(f"### Blake:\n{blake_next}\n"))
    conversation.append(("Blake", blake_next))

    charlie_next = call_charlie()
    display(Markdown(f"### Charlie:\n{charlie_next}\n"))
    conversation.append(("Charlie", charlie_next))
````
