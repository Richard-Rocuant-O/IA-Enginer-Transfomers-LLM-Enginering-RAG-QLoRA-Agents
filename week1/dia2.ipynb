## Primero - hablemos sobre la API de Chat Completions

1. La forma más simple de llamar a un LLM
2. Se llama Chat Completions porque está diciendo: "aquí hay una conversación, por favor predice qué debería venir después"
3. La API de Chat Completions fue inventada por OpenAI, pero es tan popular que se convirtió en un estándar de facto: otros proveedores como Google exponen endpoints "compatibles con OpenAI" que hablan el mismo formato.

### Empezaremos llamando a Gemini usando este formato compatible con OpenAI


```python
import os
from dotenv import load_dotenv

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

load_dotenv(override=True)
api_key = os.getenv('GOOGLE_API_KEY')

if not api_key:
    print("No se encontró ninguna API key - por favor verifica que tu archivo .env tenga GOOGLE_API_KEY definida!")
elif not api_key.startswith(("AIz", "AQ.")):
    print("Se encontró una API key, pero no empieza con AIz ni AQ.; por favor verifica que estás usando la clave correcta")
else:
    print("¡API key encontrada y se ve bien hasta ahora!")
```

## ¿Sabes qué es un Endpoint?

Si no, por favor revisa la guía de Fundamentos Técnicos en la carpeta de guías

Y, aquí hay un endpoint que podría interesarte...

```python
import requests

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}

payload = {
    "model": "gemini-3.1-flash-lite",
    "messages": [
        {"role": "user", "content": "Cuéntame un dato curioso"}]
}

payload
```

```python
response = requests.post(
    GEMINI_BASE_URL + "chat/completions",
    headers=headers,
    json=payload
)

response.json()
```

```python
response.json()["choices"][0]["message"]["content"]
```

# ¿Qué es el paquete openai?

Se conoce como una Librería Cliente de Python (Python Client Library).

No es más que un envoltorio (wrapper) alrededor de hacer exactamente esta llamada al endpoint http.

Simplemente te permite trabajar con código Python agradable en vez de estar lidiando con objetos json engorrosos.

Pero eso es todo. Es de código abierto y liviano. Algunas personas piensan que contiene el código del modelo de OpenAI - ¡no es así! Y como viste arriba, esta misma librería sirve para hablar con Gemini, Ollama o cualquier otro proveedor compatible, solo cambiando `base_url` y `api_key`.


```python
# Crear cliente de Gemini usando la librería cliente de OpenAI

from openai import OpenAI
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=api_key)

response = gemini.chat.completions.create(model="gemini-3.1-flash-lite", messages=[{"role": "user", "content": "Cuéntame un dato curioso"}])

response.choices[0].message.content

```
