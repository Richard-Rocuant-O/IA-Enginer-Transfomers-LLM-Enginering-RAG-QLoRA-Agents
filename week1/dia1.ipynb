## Tipos de prompts

Puede que ya sepas esto - pero si no, ¡te familiarizarás mucho con ello!

Modelos como GPT han sido entrenados para recibir instrucciones de una forma particular.

Esperan recibir:

**Un system prompt** que les dice qué tarea están realizando y qué tono deberían usar

**Un user prompt** -- el inicio de la conversación al que deben responder

````python
# Define nuestro system prompt - puedes experimentar con esto después, cambiando la última oración a 'Responde en markdown en español."

system_prompt = """
Eres un asistente que analiza el contenido de un sitio web,
y ofrece un resumen corto, ignorando el texto que pueda ser de navegación.
Responde en markdown. No envuelvas el markdown en un bloque de código - responde solo con el markdown.
"""
````

````python
# Define nuestro user prompt

user_prompt_prefix = """
Aquí está el contenido de un sitio web.
Proporciona un resumen corto de este sitio web.
Si incluye noticias o anuncios, resúmelos también.

"""
````

## Mensajes

La API de OpenAI espera recibir mensajes en una estructura particular.
Muchas de las otras APIs comparten esta estructura:

```python
[
    {"role": "system", "content": "aquí va el mensaje del sistema"},
    {"role": "user", "content": "aquí va el mensaje del usuario"}
]
```
Para darte un adelanto, las siguientes 2 celdas hacen una llamada bastante simple - ¡no exigiremos al todopoderoso GPT (todavía)!

````python
response = client.models.generate_content(
    model="gemini-3.6-flash",
    config=genai.types.GenerateContentConfig(system_instruction="Eres un asistente útil"),
    contents="¿Cuánto es 2 + 2?"
)
response.text
````

## Y ahora construyamos mensajes útiles para GPT-4.1-mini, usando una función

````python
# Mira cómo esta función arma el prompt para Gemini

def messages_for(website):
    return {
        "system_instruction": system_prompt,
        "contents": user_prompt_prefix + website
    }
````

````python
# Prueba esto, y luego intenta con algunos otros sitios web

messages_for(ed)
````

## Hora de juntar todo - ¡la API de Gemini es muy simple!

````python
# Y ahora: llama a la API de Gemini. ¡Te familiarizarás mucho con esto!

def summarize(url):
    website = fetch_website_contents(url)
    prompt = messages_for(website)
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        config=genai.types.GenerateContentConfig(system_instruction=prompt["system_instruction"]),
        contents=prompt["contents"]
    )
    return response.text
````

````python
summarize("https://edwarddonner.com")
````

````python
# Una función para mostrar esto de forma más agradable en la salida, usando markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
````

````python
display_summary("https://edwarddonner.com")
````
